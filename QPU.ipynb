```python
import boto3
from braket.aws import AwsDevice
from braket.devices import LocalSimulator
from braket.circuits import Circuit

# create the circuit
bell = Circuit().h(0).cnot(0, 1)
print(bell)

# instantiate the local simulator
local_sim = LocalSimulator()

# run the circuit
result = local_sim.run(bell, shots=1000).result()
counts = result.measurement_counts
print(counts)

# # get the account ID
# aws_account_id = boto3.client("sts").get_caller_identity()["Account"]
# # the name of the bucket
# my_bucket = "allstar-blackhole"
# # the name of the folder in the bucket
# my_prefix = "bracket-simulation-output"
# s3_folder = (my_bucket, my_prefix)

# choose the Rigetti hardware to run your circuit
device = AwsDevice("arn:aws:braket:us-west-1::device/qpu/rigetti/Aspen-M-3")
# execute the circuit
# define task with 1 day polling timeout
task = device.run(bell, poll_timeout_seconds=24*60*60)
print(task.result().measurement_counts) # display the results
print(task.result().measurement_counts)
```
